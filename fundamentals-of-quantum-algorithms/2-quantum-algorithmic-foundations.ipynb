Prime factorization of large integers is a very difficult task and forms the basis of RSA cryptography.
- **RSA250**: 250 decimal, 829 bit integer (prime factorization solved in 2020 using "number field seive" algorithm and leveraging distributed computing across thousands of machines worldwide)
- **RSA1024**: 309 decimal, 1024 bit integer (prime factorization unsolved - requires enormous compute times)

The time complexity of known Integer factorization classical algorithms is exponential.

## Universal Gate Sets

Default gate set for quantum circuits:

- Single-qubit unitary gates from this list: $X$, $Y$, $Z$, $H$, $S$, $S^\dagger$, $T$, and $T^\dagger$
- Controlled-NOT gates
- Single-qubit standard basis measurements

These gates are considered elementary operations. They form a universal gate set. This means that they can approximate any unitary operation on any number of qubits to any degree of accuracy desired, using circuits composed of these gates alone.

For Boolean circuits: `AND`, `OR`, `NOT`, and `FANOUT` gates.

### Computation cost as a function of gate count

- Inputs to a given computational problem can vary in length, potentially becoming arbitrarily large. Every circuit, on the other hand, has a fixed number of gates and wires. For this reason, when we think about algorithms in terms of circuits, we consider the elementary gate count as a measure of computational cost.

- An algorithm in the Boolean circuit model is described by a sequence of circuits $\{C_1, C_2, C_3, \dots\}$, where $C_n$ solves the problem for $n$-bit inputs. Here, the function $t$ representing the cost of this algorithm is defined as:
$$t(n)=size(C_n)$$

- The **Big-Oh** asymptotic notation is used to simplify computational cost analysis.

- Algorithms with a **polynomial** cost is roughly approximated as efficient. Reducing an exponential cost algorithm to polynomial complexity is the primary goal for computational advantage.
