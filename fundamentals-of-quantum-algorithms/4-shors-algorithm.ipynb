# Shor's Algorithm

## Preliminaries

- For any positive integer $N$, the set $Z_N$ is defined as:
$$Z_N = \{0,1,\ldots,N-1\}$$

- Operations such as addition and multiplication, on the elements of $Z_N$ taken modulo $N$ will always yield an element from the same set.

- The set $Z_N^*$ is defined as $Z_N^* = \{a \in Z_N:gcd(a,n) = 1\}$

- Basic fact about $Z_N^*$: Any element $a \in Z_N^*$, multiplied repeatedly to itself will eventually yield $1$ $(a^k mod N)$.

## Order Finding

**Input:** Positive integers $N$ and $a$ satisfying $\gcd(N, a) = 1$

**Output:** The smallest positive integer $r$ such that $a^r \equiv 1 \pmod{N}$

### Relating the order finding problem to phase estimation

The operation $M_a$ is defined on a system whose classical states correspond to $Z_N$, where $a \in Z_N^*$:

$$
M_a |x\rangle = |ax\rangle \quad (for-each-x \in Z_N)
$$

The operation $M_a$ is deterministic and invertible. It is described by a permutation matrix, thus it is unitary.

Eigenvector/eigenvalue pairs for $M_a$. For any choice of $j \in \{0, \dots, r - 1\}$ we have that

$$|\psi_j\rangle = \frac{1}{\sqrt{r}} \sum_{k=0}^{r-1} \omega_r^{-jk} |a^k\rangle$$

is an eigenvector of $M_a$ whose corresponding eigenvalue is $\omega_r^j$.

$$M_a |\psi_j\rangle = \omega_r^j |\psi_j\rangle$$

There are other eigenvectors of $M_a$, but we don't need to concern ourselves with them — we'll focus solely on the eigenvectors $|\psi_0\rangle, \dots, |\psi_{r-1}\rangle$.

- $n$ bit $M_a$ operations, with a precise definition:

$$
M_a|x\rangle = \begin{cases} |ax \pmod N\rangle & 0 \leq x < N \\ |x\rangle & N \leq x < 2^n \end{cases}
$$

- Each $M_a$ operation can be implemented using the algorithms for integer multiplication and division using quantum circuits, together with the methodology for reversible, garbage-free implementations of them, at cost $O(n^2)$. The circuit uses work space qubits.

- To perform the exponentiated forms of $M_a$, that is $M_a^k$ in general, compute $b = a^k$ using binary exponentiation (optimized), where $b \in Z_N^*$, and replace each $M_a^k$ with $M_b$.

- Use $m = 2lg(N)$ control qubits for a good enough approximation of $r$.

- The solution yields $\theta = \frac{y}{2^m} \approx \frac{k}{r}$.

- $r$ can be determined using the `continued fraction algorithm` (classical).

- for $|\psi\rangle$ use $|1\rangle$ which is a uniform superposition of all the possible eigenvectors of $M_a$:
$$|1\rangle = \frac{1}{\sqrt{r}} \sum_{k=0}^{r-1} |\psi_k\rangle$$

- Total cost of the order finding circuit along with required classical processing: $O(N^3)$.

## Factoring by order finding

**Probabilistic algorithm to split an odd, composite integer N that is not a prime power**

1. Randomly choose $a \in \{2, \ldots, N-1\}.$

2. Compute $d = \gcd(a, N).$

3. If $d > 1,$ then output $b = d$ and $c = \frac{N}{d}$ and stop. Otherwise, continue to the next step knowing that $a \in \mathbb{Z}_N^{*}.$

4. Let $r$ be the **order** of $a$ modulo $N$. (This is where **order finding** is required.)

5. If $r$ is even:

   1. Compute $x = a^{r/2} - 1 \pmod{N}.$

   2. Compute $d = \gcd(x, N).$

   3. If $d > 1,$ then output $b = d$ and $c = \frac{N}{d}$ and stop.

6. If this point is reached, the algorithm has failed to find a factor of $N$.
